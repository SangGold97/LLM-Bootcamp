## 2.2 MCP Server Implementation

````python
# Check and install mcp library
try:
    from mcp.server import Server
    from mcp.types import Tool, TextContent, Resource
    print("✅ MCP library is installed")
except ImportError:
    print("⚠️  Installing MCP library...")
    import subprocess
    import sys
    subprocess.check_call([sys.executable, "-m", "pip", "install", "mcp"])
    from mcp.server import Server
    from mcp.types import Tool, TextContent, Resource
    print("✅ Installation successful")
````

````python
# TODO: Create MCP Server with 4 tools
# INSTRUCTIONS:
# - Create Server instance with name "utility-server"
# - Implement 4 tool handlers: get_current_time, format_text, count_words, reverse_string
# - Register tools using @server.list_tools()
# - Implement tool execution using @server.call_tool()
#
# REFERENCE FROM THEORY:
# - server = Server("utility-server")
# - @server.call_tool() - decorator for tool execution
# - @server.list_tools() - decorator for tool discovery

from datetime import datetime
import json

# TODO: Create MCP Server
utility_server = Server("utility-server")

# TODO: Implement @server.call_tool()
# In this handler, receive tool_name and arguments
# Call the corresponding function and return the result
# 
# Example:
# if name == "get_current_time":
#     return [TextContent(type="text", text=str(datetime.now()))]
@utility_server.call_tool()
async def call_tool(name: str, args: Dict) -> Any:
    if name == "get_current_time":
        return [TextContent(type="text", text=str(datetime.now()))]
    elif name == "format_text":
        text = args.get("text", "")
        style = args.get("style", "uppercase")
        if style == "uppercase":
            formatted = text.upper()
        elif style == "lowercase":
            formatted = text.lower()
        else:
            formatted = text
        return [TextContent(type="text", text=formatted)]
    elif name == "count_words":
        text = args.get("text", "")
        word_count = len(text.split())
        return [TextContent(type="text", text=str(word_count))]
    elif name == "reverse_string":
        text = args.get("text", "")
        reversed_text = text[::-1]
        return [TextContent(type="text", text=reversed_text)]
    else:
        return [TextContent(type="text", text=f"Tool '{name}' not found.")]

# TODO: Implement @server.list_tools()
# Return list of Tool objects with:
# - name, description, inputSchema
@utility_server.list_tools()
async def list_tools() -> list:
    return [
        Tool(
            name="get_current_time",
            description="Get the current date and time",
            inputSchema={
                "type": "object",
                "properties": {},
                "required": []
            }
        ),
        Tool(
            name="format_text",
            description="Format text to uppercase or lowercase",
            inputSchema={
                "type": "object",
                "properties": {
                    "text": {"type": "string", "description": "Text to format"},
                    "style": {"type": "string", "description": "Format style: uppercase or lowercase"}
                },
                "required": ["text", "style"]
            }
        ),
        Tool(
            name="count_words",
            description="Count the number of words in a text",
            inputSchema={
                "type": "object",
                "properties": {
                    "text": {"type": "string", "description": "Text to count words"}
                },
                "required": ["text"]
            }
        ),
        Tool(
            name="reverse_string",
            description="Reverse the given string",
            inputSchema={
                "type": "object",
                "properties": {
                    "text": {"type": "string", "description": "Text to reverse"}
                },
                "required": ["text"]
            }
        )
    ]

print("Not completed - please implement MCP Server")
print("""
STEPS:
1. Create Server instance
2. Implement @call_tool() handler
3. Implement @list_tools() handler
4. Define tool implementation functions
5. Test server (call tools)
""")
````
